# Task 1- Create the Variables

```python
disease_name="cholera"
state="lagos"
total_reported_cases=850
new_cases_today=45
total_cases=total_reported_cases + new_cases_today
recovered_patients=730
death=18
available_isolation_beds=95
minimum_safe_bed_capacity=80
emergency_declared=False
print("Disease Name:",disease_name)
print("State:", state)
print("Total Reported Cases:", total_reported_cases)
print("New Cases Today:", new_cases_today)
print("Total Cases:", total_cases) 
print("Recovered Patients:", recovered_patients)
print("Death:", death)
print("Available Isolation Beds:", available_isolation_beds)
print("Minimum Safe Bed Capacity:", minimum_safe_bed_capacity)
print("Emergency Declared:", emergency_declared)
print(type(disease_name))
print(type(state))
print(type(total_reported_cases))
print(type(new_cases_today))
print(type(total_cases))
print(type(recovered_patients))
print(type(death))
print(type(available_isolation_beds))
print(type(minimum_safe_bed_capacity))
print(type(emergency_declared))

```

# Task 2- Calculate Active Cases

```python
active_cases=total_cases - recovered_patients - death
print("Active Cases:",active_cases) 
```

# Task 3- Calculate Recovery Rate

```python
recovery_rate=(recovered_patients/total_cases) *100
print("Recovery Rate:", recovery_rate)
```

# Task 4- Update Total Cases

```python
total_cases+=15
print("Total Cases:",total_cases) 
```

# Task 5- Check Bed Capacity

```python
bed_capacity=available_isolation_beds>minimum_safe_bed_capacity
print("Hospital Capacity Sufficient:",bed_capacity) 
```

# Task 6- Determine if the Outbreak is Severe

```python
active_cases>100 or death>20
print("Severe Outbreak:",active_cases>100 or death>20) 
```

# Task 7- Determine if Emergency Response is Needed

```python
outbreak_severity=True
emergency_declared=False
print("Emergency Response Needed:",outbreak_severity and not emergency_declared )
```

# Task 8- Determine if the Situation is Stable

```python
recovery_rate>80
sufficient_bed_capacity= True
print("Situation Stable:",recovery_rate>80 and sufficient_bed_capacity)
```

# Task 9- Generate an Outbreak Report

```python
print("================================")
print("     DISEASE OUTBREAK REPORT    ")
print("================================")
print("Disease Name:",disease_name)
print("State:", state)
print("Total Cases:", total_cases)
print("Recovered Patients:", recovered_patients)
print("Death:", death)
print("Active Cases:",active_cases) 
print("Recovery Rate:", recovery_rate)
print("Available Isolation Beds:", available_isolation_beds)
print("Emergency Declared:",emergency_declared)
print("Hospital Capacity Sufficient:",bed_capacity) 
print("Severe Outbreak:",active_cases>100 or death>20)
print("Emergency Response Needed:",outbreak_severity and not emergency_declared  )
print("Situation Stable:",recovery_rate>80 and sufficient_bed_capacity)
```
